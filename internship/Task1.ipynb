```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go

# Set random seed for reproducibility
np.random.seed(42)

# Generate synthetic E-commerce performance dataset
dates = pd.date_range(start='2023-01-01', periods=365, freq='D')
categories = ['Electronics', 'Clothing', 'Home & Kitchen', 'Books', 'Beauty']
regions = ['North', 'South', 'East', 'West']

data = []
for date in dates:
    for cat in categories:
        sales = np.random.normal(500, 150) + (100 if cat == 'Electronics' else 0)
        profit = sales * np.random.uniform(0.1, 0.35)
        discount = np.random.uniform(0.05, 0.3)
        region = np.random.choice(regions)
        customer_rating = np.random.uniform(3.0, 5.0)
        data.append([date, cat, region, round(sales, 2), round(profit, 2), round(discount, 2), round(customer_rating, 1)])

df = pd.DataFrame(data, columns=['Date', 'Category', 'Region', 'Sales', 'Profit', 'Discount', 'Rating'])
df['Month'] = df['Date'].dt.to_period('M').astype(str)

# ---------------------------------------------------------
# Visualization 1: Monthly Sales Trend by Category (Line Chart)
# ---------------------------------------------------------
plt.figure(figsize=(12, 6))
monthly_cat = df.groupby(['Month', 'Category'])['Sales'].sum().reset_index()
sns.lineplot(data=monthly_cat, x='Month', y='Sales', hue='Category', marker='o', linewidth=2.5)
plt.title('Monthly Sales Performance across Product Categories', fontsize=14, fontweight='bold')
plt.xlabel('Month', fontsize=12)
plt.ylabel('Total Sales ($)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('viz1_sales_trend.png')
plt.close()

# ---------------------------------------------------------
# Visualization 2: Discount vs Profitability Relationship (Scatter Plot)
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(500), x='Discount', y='Profit', hue='Category', style='Region', alpha=0.8, s=70)
sns.regplot(data=df.sample(500), x='Discount', y='Profit', scatter=False, color='black', line_kws={'linestyle': '--'})
plt.title('Impact of Discount Rates on Profit Margins', fontsize=14, fontweight='bold')
plt.xlabel('Discount Rate (%)', fontsize=12)
plt.ylabel('Profit ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('viz2_discount_vs_profit.png')
plt.close()

# ---------------------------------------------------------
# Visualization 3: Regional Sales & Profit Matrix (Heatmap)
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
pivot_tbl = df.pivot_table(values='Sales', index='Category', columns='Region', aggfunc='sum')
sns.heatmap(pivot_tbl, annot=True, fmt='.0f', cmap='YlGnBu', cbar_kws={'label': 'Total Sales ($)'})
plt.title('Regional Revenue Distribution Matrix', fontsize=14, fontweight='bold')
plt.xlabel('Region', fontsize=12)
plt.ylabel('Category', fontsize=12)
plt.tight_layout()
plt.savefig('viz3_regional_heatmap.png')
plt.close()

# ---------------------------------------------------------
# Visualization 4: Profit Margin Distribution Across Categories (Box Plot)
# ---------------------------------------------------------
plt.figure(figsize=(10, 6))
df['Profit Margin (%)'] = (df['Profit'] / df['Sales']) * 100
sns.boxplot(data=df, x='Category', y='Profit Margin (%)', palette='Set2')
plt.title('Profit Margin Variability Across Product Categories', fontsize=14, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Profit Margin (%)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('viz4_profit_margin_boxplot.png')
plt.close()

# ---------------------------------------------------------
# Visualization 5: Interactive Multi-Metric Narrative (Plotly Sunburst)
# ---------------------------------------------------------
fig = px.sunburst(
    df, 
    path=['Region', 'Category'], 
    values='Sales',
    color='Profit',
    color_continuous_scale='RdYlGn',
    title='Interactive Breakdown of Regional & Categorical Revenue & Profitability'
)
fig.write_html("viz5_interactive_sunburst.html")
%pip install -q --upgrade kaleido

fig.write_image("viz5_sunburst.png")

print("All 5 visualizations generated and saved successfully.")
```
